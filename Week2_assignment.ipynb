**Practice 2**


```python
# 3a. Create a variable called letters that contains ['a', 'b', 'c', 'd']
letters = ['a', 'b', 'c', 'd']

# 3b. What does letters[int(int('3'*2)//11)] evaluate to?
# Explanation:
# '3'*2         -> '33'
# int('33')     -> 33
# 33 // 11      -> 3
# int(3)        -> 3 (this step is redundant but harmless)
# letters[3]    -> 'd'
result = letters[int(int('3'*2)//11)]
print(result)  # Output: d

# 3c. What does letters[-1] evaluate to?
# Explanation: Negative indexing counts from the end of the list.
# -1 refers to the last element, which is 'd'.
print(letters[-1])  # Output: d

# 3d. What does letters[:2] evaluate to?
# Explanation: The slice letters[:2] starts from the beginning and goes up to,
# but does not include, index 2. It includes indices 0 and 1.
# The result is ['a', 'b'].
print(letters[:2])  # Output: ['a', 'b']
```

**Practice 4**

```python
def comma_code(items):
    items = [str(item) for item in items]

    if len(items) == 0:
        return ''
    elif len(items) == 1:
        return items[0]
    elif len(items) == 2:
        return items[0] + ' and ' + items[1]
    else:
        return ', '.join(items[:-1]) + ', and ' + items[-1]

spam = ['apples', 'bananas', 'tofu', 'cats']
print(comma_code(spam))
```
