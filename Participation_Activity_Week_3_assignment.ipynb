a. When does the code in a function execute: when the function is defined or called?

Answer: The code in a function executes when the function is called.

b. How many global scopes are there in a Python program? How many local scopes?

Answer: There is only one global scope in a Python program. A local scope is created every time a function is called.

c. What happens to variables in a local scope when the function call returns?

Answer: The local scope is destroyed, and all the variables in it are forgotten.

d. How can you prevent a program from crashing when it gets an error?

Answer: By placing the code in a try block and using an except statement to handle the error.

```python
def collatz(number):
    if number % 2 == 0:
        result = number // 2
    else:
        result = 3 * number + 1

    # Print the result on the same line, separated by a space
    print(result, end=' ')
    return result

# Main program execution with Input Validation
try:
    user_input = int(input("Enter number:\n"))

    while user_input != 1:
        user_input = collatz(user_input)

except ValueError:
    print("Error: You must enter an integer.")
```
