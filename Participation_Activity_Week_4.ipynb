# Participation Activity Week 4



## 1


```python
from pathlib import Path
import pandas as pd

DATA_FILE = Path('parsed.csv')
df = pd.read_csv(DATA_FILE)

print(f'Rows: {len(df):,}')
print(f'Columns: {len(df.columns)}')
display(df.head())
print(df.columns.tolist())
```

## 2a



```python
japan_mb = df.loc[
    df['parsed_place'].fillna('').astype(str).str.contains('Japan', case=False, na=False)
    & df['magType'].eq('mb')
].copy()

japan_95th_percentile = japan_mb['mag'].quantile(0.95)

print(f'Number of Japan mb earthquakes: {len(japan_mb)}')
print(f'95th percentile of magnitude: {japan_95th_percentile:.1f}')
```

## 2b


```python
indonesia = df.loc[
    df['parsed_place'].fillna('').astype(str).str.contains('Indonesia', case=False, na=False)
].copy()

indonesia_tsunami_count = int(indonesia['tsunami'].eq(1).sum())
indonesia_tsunami_percentage = (
    indonesia_tsunami_count / len(indonesia) * 100
)

print(f'Indonesia earthquakes: {len(indonesia)}')
print(f'Indonesia earthquakes with tsunamis: {indonesia_tsunami_count}')
print(f'Percentage with tsunamis: {indonesia_tsunami_percentage:.2f}%')
```

## 2c

```python
nevada = df.loc[
    df['parsed_place'].fillna('').astype(str).str.contains('Nevada', case=False, na=False)
].copy()

print(f'Nevada earthquakes: {len(nevada)}')
display(nevada.describe())
```

## 2d

```python
import re

ring_of_fire_locations = [
    'Bolivia', 'Chile', 'Ecuador', 'Peru', 'Costa Rica', 'Guatemala',
    'Mexico', 'Japan', 'Philippines', 'Indonesia', 'New Zealand',
    'Antarctic', 'Canada', 'Fiji', 'Alaska', 'Washington',
    'California', 'Russia', 'Taiwan', 'Tonga', 'Kermadec'
]

ring_pattern = '|'.join(re.escape(location) for location in ring_of_fire_locations)
df['is_ring_of_fire'] = df['parsed_place'].fillna('').astype(str).str.contains(
    ring_pattern, case=False, regex=True, na=False
)

new_mexico = df['parsed_place'].fillna('').astype(str).str.contains('New Mexico', case=False, na=False)
df.loc[new_mexico, 'is_ring_of_fire'] = False

display(df[['place', 'parsed_place', 'is_ring_of_fire']].head(10))
print('New Mexico rows excluded:', int(new_mexico.sum()))
```

## 2e

```python
ring_of_fire_count = int(df['is_ring_of_fire'].sum())
outside_ring_of_fire_count = int((~df['is_ring_of_fire']).sum())

print(f'Inside the Ring of Fire: {ring_of_fire_count:,}')
print(f'Outside the Ring of Fire: {outside_ring_of_fire_count:,}')
```



## 2f

```python
ring_of_fire_tsunami_count = int(
    df.loc[df['is_ring_of_fire'], 'tsunami'].eq(1).sum()
)

print(f'Tsunamis along the Ring of Fire: {ring_of_fire_tsunami_count}')
```
