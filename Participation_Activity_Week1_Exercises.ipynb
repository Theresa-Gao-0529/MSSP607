## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

To calculate the mean without imports, we sum all the values and divide by the total number of values.

```python
total_salary = sum(salaries)
number_of_salaries = len(salaries)
mean_salary = total_salary / number_of_salaries

print(f"Mean salary: {mean_salary}")
```

### median

To calculate the median without imports, we first sort the list. If the number of elements is odd, the median is the middle element. If the number of elements is even, the median is the average of the two middle elements.

```python
sorted_salaries = sorted(salaries)
n = len(sorted_salaries)

if n % 2 == 1:
    # Odd number of elements
    median_salary = sorted_salaries[n // 2]
else:
    # Even number of elements
    mid1 = sorted_salaries[n // 2 - 1]
    mid2 = sorted_salaries[n // 2]
    median_salary = (mid1 + mid2) / 2

print(f"Median salary: {median_salary}")
```

### mode

To calculate the mode without imports, we count the occurrences of each element and then find the element(s) with the highest frequency.

```python
salary_counts = {}
for salary in salaries:
    salary_counts[salary] = salary_counts.get(salary, 0) + 1

# Find the maximum frequency
max_frequency = 0
if salary_counts:
    max_frequency = max(salary_counts.values())

# Find all salaries that have the maximum frequency
mode_salaries = [salary for salary, count in salary_counts.items() if count == max_frequency]

print(f"Mode salary(ies): {mode_salaries}")
```

### sample variance
Remember to use Bessel's correction.

To calculate the sample variance with Bessel's correction without imports, we first calculate the mean. Then, we sum the squared differences between each data point and the mean, and divide by `n-1` (where `n` is the number of data points).

```python
# Calculate the mean first (re-calculating for completeness within this block)
total_salary = sum(salaries)
number_of_salaries = len(salaries)
mean_salary_for_variance = total_salary / number_of_salaries

# Calculate the sum of squared differences from the mean
sum_squared_diffs = 0
for salary in salaries:
    sum_squared_diffs += (salary - mean_salary_for_variance) ** 2

# Apply Bessel's correction (divide by n-1 for sample variance)
sample_variance = sum_squared_diffs / (number_of_salaries - 1)

print(f"Sample variance: {sample_variance}")
```

### sample standard deviation
Remember to use Bessel's correction.

The sample standard deviation is the square root of the sample variance. We can use the `** 0.5` operator or `math.sqrt()` for this, but since we are avoiding imports, we'll use `** 0.5`.

```python
# Calculate the sample standard deviation from the previously calculated sample variance
sample_std_dev = sample_variance ** 0.5

print(f"Sample standard deviation: {sample_std_dev}")
```

## Exercise 6: Calculating more statistics
### range

The range of a dataset is the difference between the maximum and minimum values. To calculate this without imports, we can iterate through the list to find the highest and lowest salaries.

```python
max_salary = salaries[0]
min_salary = salaries[0]

for salary in salaries:
    if salary > max_salary:
        max_salary = salary
    if salary < min_salary:
        min_salary = salary

salary_range = max_salary - min_salary

print(f"Maximum salary: {max_salary}")
print(f"Minimum salary: {min_salary}")
print(f"Range of salaries: {salary_range}")
```

### coefficient of variation

The coefficient of variation (CV) is a measure of relative variability. It is the ratio of the standard deviation to the mean, often expressed as a percentage. It helps to compare the degree of variation between data series, even if their means are drastically different.

```python
coefficient_of_variation = (sample_std_dev / mean_salary) * 100 # Express as a percentage

print(f"Coefficient of Variation: {coefficient_of_variation:.2f}%")
```

### interquartile range

The Interquartile Range (IQR) is a measure of statistical dispersion, specifically the difference between the 75th percentile (third quartile, Q3) and the 25th percentile (first quartile, Q1) of the data. To calculate it without imports, we first need to ensure the list is sorted. Then, Q1 is the median of the lower half of the data, and Q3 is the median of the upper half.

```python
# The 'salaries' list is already sorted into 'sorted_salaries' from previous calculations
# n is the length of sorted_salaries, which is 100

# Calculate Q1 (25th percentile)
# For an even number of elements (n=100), the lower half has 50 elements (indices 0 to 49).
# Q1 is the median of this lower half, which is the average of its 25th and 26th elements (original indices 24 and 25).
idx_q1_1 = (n // 4) - 1 # 25th element (index 24)
idx_q1_2 = (n // 4)     # 26th element (index 25)
Q1 = (sorted_salaries[idx_q1_1] + sorted_salaries[idx_q1_2]) / 2

# Calculate Q3 (75th percentile)
# The upper half has 50 elements (indices 50 to 99).
# Q3 is the median of this upper half, which is the average of its 25th and 26th elements (original indices 50+24 and 50+25, i.e., 74 and 75).
idx_q3_1 = (3 * n // 4) - 1 # 75th element (index 74)
idx_q3_2 = (3 * n // 4)     # 76th element (index 75)
Q3 = (sorted_salaries[idx_q3_1] + sorted_salaries[idx_q3_2]) / 2

interquartile_range = Q3 - Q1

print(f"First Quartile (Q1): {Q1}")
print(f"Third Quartile (Q3): {Q3}")
print(f"Interquartile Range (IQR): {interquartile_range}")
```

### quartile coefficent of dispersion

The Quartile Coefficient of Dispersion (QCD) is a measure of relative dispersion, calculated as the ratio of the interquartile range to the sum of the first and third quartiles. It is given by the formula: `(Q3 - Q1) / (Q3 + Q1)`.

```python
quartile_coefficient_of_dispersion = (Q3 - Q1) / (Q3 + Q1)

print(f"Quartile Coefficient of Dispersion: {quartile_coefficient_of_dispersion:.4f}")
```
