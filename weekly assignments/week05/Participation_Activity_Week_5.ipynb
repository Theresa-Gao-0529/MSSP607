## 1. Import pandas

```python
import pandas as pd
```

## 2. Read each file into a DataFrame

```python
aapl = pd.read_csv('/content/MSSP6070/Weekly Assignments/Week05/Data/aapl.csv')
amzn = pd.read_csv('/content/MSSP6070/Weekly Assignments/Week05/Data/amzn.csv')
fb = pd.read_csv('/content/MSSP6070/Weekly Assignments/Week05/Data/fb.csv')
goog = pd.read_csv('/content/MSSP6070/Weekly Assignments/Week05/Data/goog.csv')
nflx = pd.read_csv('/content/MSSP6070/Weekly Assignments/Week05/Data/nflx.csv')

aapl.head()
```

## 3. Add the ticker column

```python
aapl['ticker'] = 'AAPL'
amzn['ticker'] = 'AMZN'
fb['ticker'] = 'FB'
goog['ticker'] = 'GOOG'
nflx['ticker'] = 'NFLX'

aapl.head()
```

## 4. Append the five DataFrames

```python
faang = pd.concat([aapl, amzn, fb, goog, nflx], ignore_index=True)
faang.head()
```

## 5. Save the combined data as `faanf.csv`

```python
faang.to_csv('/content/MSSP6070/Weekly Assignments/Week05/Data/faanf.csv', index=False)
print('Saved Data/faanf.csv')
```

## 6. Convert data types and sort

```python
faang['date'] = pd.to_datetime(faang['date'])
faang['volume'] = faang['volume'].astype(int)
faang = faang.sort_values(['date', 'ticker']).reset_index(drop=True)
faang.dtypes
```

```python
faang.head(10)
```

## 7. Find the seven rows with the highest volume

```python
top7_volume = faang.nlargest(7, 'volume')
top7_volume
```

## 8. Use `melt()` with `date` and `ticker` as identifiers

```python
faang_melted = faang.melt(id_vars=['date', 'ticker'])
faang_melted.head(10)
```

## 9. Final checks

```python
print('Combined DataFrame shape:', faang.shape)
print('\nTicker counts:')
print(faang['ticker'].value_counts())
print('\nTop 7 volume rows:')
display(top7_volume)
print('\nMelted DataFrame shape:', faang_melted.shape)
```
